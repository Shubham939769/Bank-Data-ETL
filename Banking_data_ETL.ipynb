```python

import re
import pandas as pd

def clean_company_name(name):

    # Handle missing values
    if pd.isna(name):
        return pd.NA

    name = str(name).strip().upper()

    # Remove unwanted leading special characters
    name = re.sub(r'^[^A-Z0-9]+', '', name)

    # Remove 1-2 digit prefixes, except names starting with 0 TO
    name = re.sub(
        r'^(?!0\s+TO\b)\d{1,2}(?=\s|[A-Z]|[._-])[\s._-]*',
        '',
        name
    )

    # Remove unwanted special characters
    # Preserve &, brackets, apostrophes, hyphens and slashes
    name = re.sub(r"[^A-Z0-9\s&().,'/-]", ' ', name)

    # Remove extra spaces
    name = re.sub(r'\s+', ' ', name).strip()

    return name
```

```python
final_df['Cleaned_Company_Name'] = (
    final_df['Company Name'].apply(clean_company_name)
)

print(final_df[
    ['Company Name', 'Cleaned_Company_Name']
].head(20))
```

```python

# Statuses to remove
remove_status = [
    'Strike Off',
    'Dissolved',
    'Dissolved (Liquidated)',
    'Dissolved under section 54',
    'Dissolved under section 59(8)',
    'Inactive',
    'Vanished'
]

# Count before filtering
before = len(final_df)

# Remove unwanted statuses from Company Status column
final_df = final_df[
    ~final_df['Company Status']
    .astype('string')
    .str.strip()
    .str.casefold()
    .isin([s.casefold() for s in remove_status])
].copy()

# Count after filtering
print("Records before:", before)
print("Records removed:", before - len(final_df))
print("Records remaining:", len(final_df))

# Verify
print(final_df['Company Status'].value_counts())
```

```python

import pandas as pd
import os

# Folder containing your dataset
folder_path = r"C:\Users\Shree\Desktop\python code\zuperr_company"

# Define four separate status files
status_files = {
    "Under process of striking off": "Under_Process_Striking_Off.csv",
    "Strike Off-Awaiting Publication": "Strike_Off_Awaiting_Publication.csv",
    "Amalgamated": "Amalgamated_Companies.csv",
    "Unclassified": "Unclassified_Companies.csv"
}

# Track records to remove from original dataset
remove_mask = pd.Series(False, index=final_df.index)

# Create four separate files
for status, filename in status_files.items():

    mask = (
        final_df['Company Status']
        .astype('string')
        .str.strip()
        .str.casefold()
        .eq(status.casefold())
    )

    # Extract matching records
    status_df = final_df.loc[mask].copy()

    # Save separate CSV
    output_file = os.path.join(folder_path, filename)
    status_df.to_csv(output_file, index=False)

    # Track extracted records
    remove_mask = remove_mask | mask

    print(f"{status}: {len(status_df)} records saved")

# Remove all four statuses from original dataset
final_df = final_df.loc[~remove_mask].copy()

# Save updated master dataset
final_path = os.path.join(
    folder_path,
    "Company_Master_Final.csv"
)

final_df.to_csv(final_path, index=False)

print("\nAll four files created successfully!")
print("Remaining master records:", len(final_df))
print("Master file:", final_path)
```
